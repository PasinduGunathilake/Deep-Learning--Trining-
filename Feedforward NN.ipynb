1.Calculate hidden layer input and output

Define input function


```python
import numpy as np

def calc_node_input(weight,input,bias):
    #dot product of 1d array
    hidden_input = np.dot(weight,input)+ bias
    return hidden_input
```

Define Activation functions

```python
#output function ReLU

def relu(input_value):
    return np.maximum(0,input_value)

#output function sigmoid

def sigmoid(input_value):
    return 1/(1+ np.exp(-input_value))

#output function TanH
def tanh(input_value):
    return ((np.exp(input_value) - np.exp(-input_value)) / (np.exp(input_value) + np.exp(-input_value)))
```

Define cost function

```python
def calc_node_output(node_input, activation_function):
    if (activation_function == 'relu'):
        return relu(node_input)
    elif (activation_function == 'sigmoid'):
        return sigmoid(node_input)
    elif (activation_function == 'tanh'):
        return tanh(node_input)
    else:
        print('Enter valid Activation function')
```

Define inputs, weights and biases

```python
#inputs
x1= 0.3
x2= 0.5

#weights of nets
w1 =0.1
w2 =0.2
w3 =0.3
w4 =0.4
w5 =0.5
w6 =0.6
w7 =0.7
w8 =0.8
w9 =0.9
w10 =0.1
w11 =0.2
w12 =0.3

#biases of 2 hidden layers
b1= 0.5
b2= 0.5



#Expected outputs

o1 =0
o2 =1
```

CALCULATION

First Hidden Layer (can use different different activation functions for layer)


```python
w_h11 =[w1,w3]
x_h11 =[x1,x2]
b_h11 = b1

h11_input = calc_node_input(w_h11,x_h11,b_h11)
print ('h11 input:',h11_input)
h11_output = calc_node_output(h11_input,'sigmoid')
print('h11 output:',h11_output)
```

```python
w_h12 = [w2,w4]
x_h12 =[x1,x2]
b_h12 = b1
h12_input = calc_node_input(w_h12,x_h12,b_h12)
print ('h12 input:',h12_input)
h12_output = calc_node_output(h12_input,'sigmoid')
print('h12 output:',h12_output)
```

Second Hidden Layer

```python
w_h21 = [w5,w7]
x_h21 = [h11_output,h12_output]
b_h21 = b2
h21_input = calc_node_input(w_h21,x_h21, b_h21)
print('h21 input:',h21_input)
h21_output = calc_node_output(h21_input, 'sigmoid')
print('h21 output:',h21_output)
```

```python
w_h22 =[w6,w8]
x_h22 =[h11_output,h12_output]
b_h22 = b2
h22_input= calc_node_input(w_h22,x_h22,b_h22)
print('h22 input:',h22_input)
h22_output= calc_node_output(h22_input,'sigmoid')
print('h22 output',h22_output)
```

Final Layer / Output Layer

```python
w_y1= [w9,w11]
x_y1= [h21_output,h22_output]
y1_input = calc_node_input(w_y1,x_y1,0)
print('y1 input:',y1_input)
y1_output = calc_node_output(y1_input,'sigmoid')
print('y1 output:',y1_output)
```

```python
w_y2 = [w10,w12]
x_y2 = [h21_output, h22_output]
y2_input = calc_node_input(w_y2,x_y2,0)
print('y2 input: ',y2_input)
y2_output = calc_node_output(y2_input,'sigmoid')
print('y2 output:',y2_output)
```

2.Error calculation


```python
#Mean Squad Error
import math

def mse(n,result1,result2,expected_result1,expected_result2):
    error = 1/n*(math.pow(result1-expected_result1,2)+(math.pow(result2-expected_result2,2)))
    return error
```

CALCULATION

```python
print(mse(2,y1_output,y2_output,o1,o2))
```
